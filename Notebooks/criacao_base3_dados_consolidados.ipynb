##CONSOLIDAÇÃO DAS BASES

- Com a análise das duas fontes concluída, será criada uma terceira base consolidada para utilização nas análises de People Analytics.

- A Base 3 será construída a partir da relação entre a Base 1 e a Base 2 por meio do campo `ID_Colaborador`.

- A consolidação permitirá reunir, em uma única estrutura, as informações cadastrais e organizacionais dos colaboradores e as informações relacionadas às suas admissões e desligamentos.
<br>
OBJETIVO DA BASE 3

- A Base 3 terá como objetivo disponibilizar uma visão consolidada dos colaboradores, servindo como principal fonte para as análises e indicadores das próximas etapas do projeto.

- A consolidação não terá como objetivo eliminar as inconsistências identificadas nas fontes originais.

- Os dados serão preservados e as situações identificadas durante a análise de qualidade serão sinalizadas por meio de campos auxiliares.
<br>

CAMPOS DERIVADOS

Além das informações provenientes das Bases 1 e 2, a Base 3 contará com campos derivados para facilitar as análises posteriores.

<br>
STATUS - O campo `Status` será utilizado para classificar o vínculo do colaborador. A regra inicial será:

- Data de desligamento preenchida → **Desligado**
- Data de desligamento não preenchida + motivo informado → **A validar**
- Data de desligamento não preenchida + motivo não informado → **Ativo**

>A classificação **A validar** permite identificar registros em que existe uma indicação de desligamento, mas não há uma data registrada para confirmar a situação.

<br>
FLAG DE QUALIDADE -
O campo `Flag_Qualidade` será utilizado para sinalizar inconsistências identificadas nos registros.

A flag permitirá preservar os dados originais e, ao mesmo tempo, identificar situações que possam exigir atenção durante as análises. Entre as situações identificadas estão:

- motivo de desligamento sem data;
- data de desligamento sem motivo;
- desligamento anterior à admissão;
- admissão e desligamento na mesma data.

<br>
TEMPO DE CASA - O campo `Tempo_de_Casa` será calculado a partir da diferença entre a data de admissão e a data de referência correspondente.

- Para colaboradores desligados, será considerada a data de desligamento.

- Para colaboradores ativos, será considerada a data atual.

>O resultado será apresentado em anos e meses, facilitando a interpretação dos dados nas análises de People Analytics.

<br>

##DECISÃO SOBRE O TRATAMENTO DOS DADOS

Neste momento, não serão realizadas correções automáticas ou exclusões dos registros identificados como inconsistentes. A decisão considera que uma inconsistência pode representar tanto um erro de preenchimento quanto uma situação que necessita de investigação. Dessa forma, a estratégia adotada será:

1. Preservar os dados originais;
2. Consolidar as informações das duas fontes;
3. Sinalizar as inconsistências encontradas;
4. Criar campos derivados que facilitem a análise;
5. Avaliar posteriormente o impacto dessas situações nos indicadores de People Analytics.

> O objetivo é evitar que o processo de tratamento elimine informações potencialmente relevantes antes que seu impacto analítico seja compreendido.

```python
# CONSOLIDAÇÃO DAS BASES

df_base3 = pd.merge(
    df_colaboradores,
    df_desligamentos,
    on="ID_Colaborador",
    how="left"
)

df_base3.head(10)
```

```python
# Verifica a estrutura da Base 3
df_base3.info()
```

```python
# Verifica a quantidade de registros e colunas
df_base3.shape
```

```python
# Cria o Status do colaborador
df_base3["Status"] = "Ativo"

df_base3.loc[
    df_base3["Data_Desligamento"].notnull(),
    "Status"
] = "Desligado"

df_base3.loc[
    df_base3["Data_Desligamento"].isnull()
    & df_base3["Motivo_Desligamento"].notnull(),
    "Status"
] = "A validar"
```

```python
# Verifica a quantidade de colaboradores por Status
df_base3["Status"].value_counts()
```

```python
# Cria a coluna Flag_Qualidade
df_base3["Flag_Qualidade"] = "OK"

# Motivo informado sem data de desligamento
df_base3.loc[
    df_base3["Data_Desligamento"].isnull()
    & df_base3["Motivo_Desligamento"].notnull(),
    "Flag_Qualidade"
] = "Motivo sem data"

# Data de desligamento informada sem motivo
df_base3.loc[
    df_base3["Data_Desligamento"].notnull()
    & df_base3["Motivo_Desligamento"].isnull(),
    "Flag_Qualidade"
] = "Data sem motivo"

# Desligamento anterior à admissão
df_base3.loc[
    df_base3["Data_Desligamento"] < df_base3["Data_Admissao"],
    "Flag_Qualidade"
] = "Desligamento anterior à admissão"

# Admissão e desligamento na mesma data
df_base3.loc[
    df_base3["Data_Desligamento"] == df_base3["Data_Admissao"],
    "Flag_Qualidade"
] = "Admissão e desligamento na mesma data"
```

```python
# Verifica a quantidade de registros por Flag_Qualidade
df_base3["Flag_Qualidade"].value_counts()
```

VERIFICAÇÃO DAS INCONSISTÊNCIAS

Após a criação da `Flag_Qualidade`, foi realizada uma verificação específica dos registros que possuem data de desligamento, mas não possuem motivo informado.

>Essa verificação tem como objetivo identificar os registros envolvidos e avaliar se essas inconsistências também apresentam outras situações de qualidade, evitando que uma ocorrência seja sobrescrita por outra na criação da flag.

```python
# Verifica registros com data de desligamento sem motivo
df_base3[
    df_base3["Data_Desligamento"].notnull()
    & df_base3["Motivo_Desligamento"].isnull()
][
    ["ID_Colaborador", "Data_Admissao", "Data_Desligamento", "Flag_Qualidade"]
]
```

ANÁLISE DA FLAG DE QUALIDADE

Durante a validação da `Flag_Qualidade`, foi identificado que algumas inconsistências podem ocorrer simultaneamente no mesmo registro.

- No conjunto analisado, um colaborador apresentou simultaneamente ausência de motivo de desligamento e admissão e desligamento na mesma data.

A lógica inicial da flag utilizava condições sequenciais, fazendo com que uma inconsistência pudesse sobrescrever outra. Dessa forma, a estrutura da flag será ajustada para permitir o registro de múltiplas inconsistências no mesmo colaborador.

> Uma mesma linha pode apresentar mais de um problema de qualidade. Portanto, a identificação das inconsistências não deve eliminar informações já encontradas durante a análise.

```python
# Recria a Flag_Qualidade permitindo múltiplas inconsistências
df_base3["Flag_Qualidade"] = ""

# Motivo informado sem data de desligamento
df_base3.loc[
    df_base3["Data_Desligamento"].isnull()
    & df_base3["Motivo_Desligamento"].notnull(),
    "Flag_Qualidade"
] += "Motivo sem data; "

# Data de desligamento informada sem motivo
df_base3.loc[
    df_base3["Data_Desligamento"].notnull()
    & df_base3["Motivo_Desligamento"].isnull(),
    "Flag_Qualidade"
] += "Data sem motivo; "

# Desligamento anterior à admissão
df_base3.loc[
    df_base3["Data_Desligamento"] < df_base3["Data_Admissao"],
    "Flag_Qualidade"
] += "Desligamento anterior à admissão; "

# Admissão e desligamento na mesma data
df_base3.loc[
    df_base3["Data_Desligamento"] == df_base3["Data_Admissao"],
    "Flag_Qualidade"
] += "Admissão e desligamento na mesma data; "

# Registros sem inconsistências
df_base3.loc[
    df_base3["Flag_Qualidade"] == "",
    "Flag_Qualidade"
] = "OK"

# Remove o "; " do final das flags
df_base3["Flag_Qualidade"] = (
    df_base3["Flag_Qualidade"]
    .str.rstrip("; ")
)
```

CORREÇÃO DA FLAG DE QUALIDADE

Inicialmente, a coluna é criada vazia. As inconsistências são adicionadas à flag conforme cada condição é identificada.

- O uso de `+=` permite acumular mais de uma inconsistência no mesmo registro, evitando que uma nova condição sobrescreva uma inconsistência identificada anteriormente.

>Ao final, os registros que não apresentaram nenhuma inconsistência recebem a classificação `OK`.

```python
# Verifica resultado
df_base3["Flag_Qualidade"].value_counts()
```

```python
# Cria a data de referência para o cálculo
data_referencia = pd.Timestamp.today()

# Calcula o tempo de casa em dias
df_base3["Tempo_de_Casa_Dias"] = (
    df_base3["Data_Desligamento"].fillna(data_referencia)
    - df_base3["Data_Admissao"]
).dt.days
```

```python
# Verifica resultado
df_base3["Data_Desligamento"].fillna(data_referencia)
```

```python
# Verifica o resultado do tempo de casa
df_base3["Tempo_de_Casa_Dias"].head(10)
```

```python
# Converte o tempo de casa para anos e meses
def calcular_tempo_de_casa(dias):
    anos = dias // 365
    meses = (dias % 365) // 30

    if anos > 0 and meses > 0:
        return f"{anos} anos, {meses} meses"
    elif anos > 0:
        return f"{anos} anos"
    else:
        return f"{meses} meses"

df_base3["Tempo_de_Casa"] = (
    df_base3["Tempo_de_Casa_Dias"]
    .apply(calcular_tempo_de_casa)
)
```

```python
# Verifica o tempo de casa calculado
df_base3[
    [
        "ID_Colaborador",
        "Data_Admissao",
        "Data_Desligamento",
        "Status",
        "Tempo_de_Casa"
    ]
].head(10)
```

```python
# Corrige a apresentação de anos e meses
def calcular_tempo_de_casa(dias):
    anos = dias // 365
    meses = (dias % 365) // 30

    partes = []

    if anos > 0:
        partes.append(f"{anos} ano" if anos == 1 else f"{anos} anos")

    if meses > 0:
        partes.append(f"{meses} mês" if meses == 1 else f"{meses} meses")

    return ", ".join(partes)

df_base3["Tempo_de_Casa"] = (
    df_base3["Tempo_de_Casa_Dias"]
    .apply(calcular_tempo_de_casa)
)
```

```python
# Verifica o resultado final do tempo de casa
print(df_base3["Tempo_de_Casa"].head(10))
```
