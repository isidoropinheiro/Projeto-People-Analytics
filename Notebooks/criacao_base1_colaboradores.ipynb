IMPORTAÇÃO DAS BIBLIOTECAS


*   Pandas: trabalha com dados estruturados, principalmente tabelas.
*   Numpy: trabalha com números, arrays e cálculos matemáticos de forma eficiente.



```python
import pandas as pd
import numpy as np
```

CONFIGURAÇÕES


*   Usaremos o módulo `np.random` para gerar dados aleatórios que comporão a base, respeitando a quantidade de colaboradores definida.

> A função `seed (x)` define a semente para a geração dos números pseudoaleatórios. O x pode ser qualquer número. Seeds diferentes geram sequências diferentes, permitindo criar bases aleatórias distintas e reproduzir cada uma posteriormente.

```python
np.random.seed(40)

QUANTIDADE_COLABORADORES = 500
```

BASE DE DADOS

Para a criação da base de dados, teremos:
*   Listas Obrigatórias para atender os itens: Cargo, Senioridade, Região, Area e Gestor.
*   Funções para atender: ID Colaboradores, Salario.

> AA lógica para os Gestores é que, na base de 500 pessoas, tenhamos 30 Gestores, gerando assim times mais perto da realidade do dia a dia.





```python
#Listas

cargos = [
    "Analista",
    "Assistente",
    "Coordenador",
    "Especialista",
    "Desenvolvedor",
    "Engenheiro de Dados",
    "Gerente",
    "Consultor"
]

senioridades = [
    "Estágio",
    "Júnior",
    "Pleno",
    "Sênior",
    "Especialista"
]

regiao = [
    "SP",
    "RS",
    "PR",
    "RJ",
    "MG",
    "BA",
    "DF",
    "SC"
]

areas = [
    "Tecnologia",
    "Recursos Humanos",
    "Financeiro",
    "Marketing",
    "Comercial",
    "Operações",
    "Jurídico",
    "Produto",
    "Dados"
]

gestores = [
    f"Gestor_{i:03d}"
    for i in range(1, 31)
]
```

CRIANDO O DATAFRAME (E AS FUNÇÕES)

Depois de estipulados alguns dos valores que serão usados como dados, criamos o DataFrame, que será a nossa base de dados. Tendo as colunas:

*   `ID_Colaborador` cria um identificador para cada colaborador da base, que no total deverão somar 500 colaboradores apenas.

```python
df = pd.DataFrame()

# ID
df["ID_Colaborador"] = [
    f"COL{i:05d}"
    for i in range(1, QUANTIDADE_COLABORADORES + 1)
]
```

DADOS ORGANIZACIONAIS
*   `np.random.choice` escolhe **dentro das listas disponibilizadas** o dado que será utilizado de forma aleatória.
>Para a lógica das senioridades, utilizamos as porcentagens de probabilidade para simular times coesos dentro da realidade de times de desenvolvimento.
*   **As probabilidades devem somar 1 (100%)** - Por isso: 0.08 + 0.30 + 0.35 + 0.22 + 0.05 = 1.00

```python
df["Area"] = np.random.choice(
    areas,
    QUANTIDADE_COLABORADORES
)

df["Cargo"] = np.random.choice(
    cargos,
    QUANTIDADE_COLABORADORES
)

df["Senioridade"] = np.random.choice(
    senioridades,
    QUANTIDADE_COLABORADORES,
    p=[
        0.08,  # Estágio
        0.30,  # Júnior
        0.35,  # Pleno
        0.22,  # Sênior
        0.05   # Especialista
    ]
)

df["Gestor"] = np.random.choice(
    gestores,
    QUANTIDADE_COLABORADORES
)

df["Regiao"] = np.random.choice(
    regiao,
    QUANTIDADE_COLABORADORES
)
```

DADOS SALARIAIS
* `faixas_salariais` contém os valores mínimo e máximo para cada senioridade.
* `gerar_salario(senioridade)` recebe a senioridade e utiliza a faixa correspondente.
* `np.random.uniform(minimo, maximo)` gera um valor aleatório dentro daquela faixa, arrendodando o valor para duas casas decimais.
* `.apply(gerar_salario)` aplica a função para cada valor da coluna Senioridade e gera o salário correspondente.
* `df["Salario"]` coloca os valores gerados em uma nova coluna do DataFrame.



```python
faixas_salariais = {
    "Estágio": (1200, 1800),
    "Júnior": (2500, 4500),
    "Pleno": (4500, 7500),
    "Sênior": (7500, 12000),
    "Especialista": (11000, 18000)
}

def gerar_salario(senioridade):
    minimo, maximo = faixas_salariais[senioridade]

    return round(
        np.random.uniform(minimo, maximo),
        2
    )

df["Salario"] = df["Senioridade"].apply(
    gerar_salario
)
```

EXPORTANDO PARA EXCEL

A exportação do documento final será feita de forma simplificada, sendo gerada e armazenada dentro das pastas do próprio notebook. Posteriormente, o arquivo será enviado manualmente para o repositório do projeto.


```python
nome_arquivo = "base1_colaboradores.xlsx"

df.to_excel(
    nome_arquivo,
    index=False,
    sheet_name="Colaboradores"
)

print(
    f"Base criada com sucesso: {nome_arquivo}"
)
```
